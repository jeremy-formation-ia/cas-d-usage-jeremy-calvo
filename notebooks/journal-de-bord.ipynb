# Journal de bord — Marketing bancaire

Auteur : Jérémy Calvo


## Jour 1 — 29/09/2026

### Étapes / Actions réalisées

- Reprise du rendu depuis le début.
- Mise en place de l'environnement de travail (Python 3.11) et du fichier de dépendances.
- Création de deux notebooks : le notebook principal et ce journal de bord.
- Traçabilité : source, hash et dimensions du jeu de données, versions des librairies, commit Git.
- Rédaction du cadrage : contexte client, problème IA, bénéficiaires, critères de succès, risques éthiques et réglementaires.
- Identification et acquisition des données : sources, chargement du fichier, dictionnaire des variables et synthèse.
- Exploration des données : qualité, distributions, équilibre de la cible, corrélations et écarts entre groupes.

### Observations / Difficultés

- Il a fallu aligner l'environnement sur Python 3.11 pour rester cohérent avec les modules de formation.
- Le chemin vers le jeu de données a dû être rendu robuste, le notebook et les données n'étant pas dans le même dossier.
- Qualification des variables : distinction entre donnée personnelle, catégorie particulière (article 9) et variable sensible pour le ciblage commercial. L'âge est un bon exemple de nuance, personnel et utile à la banque mais aussi critère de discrimination possible.
- Le jeu ne contient aucun identifiant client : impossible de savoir si les doublons sont le même client ou deux clients au profil identique. Décision de les supprimer, leur part étant négligeable.
- pdays contient une sentinelle (999 = jamais recontacté) qui ne doit pas être lue comme une durée. Remplacée par un indicateur binaire.
- Les taux de souscription varient fortement selon l'âge, la profession, la situation familiale et l'éducation. Certains groupes sont très petits, ce qui rend les écarts peu fiables.

### Prochaines étapes

- Préparer les données : split, recette de prétraitement, scénarios de variables.
- Compléter la date d'extraction du jeu de données.


## Jour 3 — 03/10/2026

### Étapes / Actions réalisées

- Analyse comparative des scénarios et arbitrage final.
- Ajout d'une analyse du seuil de décision (fonction threshold_analysis dans src/modeling.py) et de tests associés.
- Choix du modèle et des variables retenus, justification par les chiffres.
- Interprétation pour le client : importance des variables par permutation, analyse des erreurs et stratégies de repli, message métier.

### Observations / Difficultés

- Le scénario 4 (sans historique de campagne) est retenu : sa performance est quasi identique au scénario 3, pour un périmètre de variables plus réduit. La plupart des clients n'ont jamais été contactés, l'historique apporte donc peu.
- Le choix du seuil est un arbitrage métier : préférer les faux positifs (contacter un prospect qui refusera) aux faux négatifs (manquer un client qui aurait souscrit). Seuil retenu : 0,30.
- Le seuil a été choisi en validation croisée sur le train (probabilités out-of-fold), puis confirmé sur le jeu de test scellé.
- Le modèle retenu ne fournit pas directement d'importance des variables (contrairement à une forêt) ; l'importance a été mesurée par permutation.
- Les variables qui pèsent le plus sont le contexte macro-économique et la période de contact, peu les données personnelles.

### Prochaines étapes

- Industrialiser la solution : persistance du modèle, API, architecture cible, conteneurisation.
- Compléter la date d'extraction du jeu de données.

## Jour 4 — 04/10/2026

### Étapes / Actions réalisées

- Industrialisation de la solution.
- Persistance du modèle retenu (pipeline complet + métadonnées).
- Développement d'une API FastAPI (santé, informations, prédiction) avec validation des entrées, gestion des erreurs et journalisation.
- Interface Streamlit pour saisir un profil client et consulter la probabilité.
- Conteneurisation (Docker, docker-compose) et intégration continue (GitHub Actions) avec contract test.
- Journal des expériences (experiments.md).

### Observations / Difficultés

- Le modèle est persisté sous forme de pipeline complet, pour que le prétraitement appliqué en production soit identique à celui de l'entraînement.
- Le schéma d'entrée de l'API est aligné sur les variables du scénario retenu (S4), avec des bornes de validation issues de l'exploration des données.
- L'API distingue la décision de contact de l'abstention : une probabilité proche du seuil est renvoyée au conseiller.
- Le choix d'un hébergement on-premise (ou cloud souverain) découle du caractère personnel des données.

### Prochaines étapes

- Suivi en production : métriques techniques et métier, dérive, réentraînement, boucle de rétroaction.
- Compléter la date d'extraction du jeu de données.
